```python
# criado vendas de dados mensais
vendas_mensais = pd.Series([1000, 2000, 3000, 4000, 5000], index=['jan', 'fev', 'mar', 'abr', 'mai'])
```

```python
vendas_mensais
```

```python
#calc e a media de vendas_mensais.mean()

media_vendas = vendas_mensais.mean()
media_vendas
```

```python
mes_maior_venda = vendas_mensais.idxmax()
mes_maior_venda
```

```python
exibir_vendas = vendas_mensais[vendas_mensais > media_vendas]
exibir_vendas
```
