```python
dados_notas = {
    'nome': ['João', 'Maria', 'Pedro', 'Ana'],
    'Matematica': [8.5, 9.0, 7.5, 8.0],
    'Portugues': [7.0, 8.5, 9.0, 7.5],
    'Historia': [9.0, 8.0, 7.5, 8.5],
    'Fisica': [8.0, 7.5, 9.0, 8.5],}

df_notas = pd.DataFrame(dados_notas)
df_notas
```

```python
#Calculando a média das notas
df_notas['Media'] = df_notas[['Matematica', 'Portugues', 'Historia', 'Fisica']].mean(axis=1)

df_notas
```

```python
df_notas[['nome', 'Media']].sort_values(by='Media', ascending=False)
```
