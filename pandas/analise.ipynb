```python
import pandas as pd
```

```python
dados_vendas = {
  "Produto": ['camisa', 'calça', 'sapato', 'meia', 'jaqueta', 'saia'],
  "Preço": [50, 100, 150, 20, 200, 80],
  "Quantidade": [10, 5, 8, 20, 3, 7],
  "Mes": ['Janeiro', 'Fevereiro', 'Março', 'Abril', 'Maio', 'Junho']

}
```

```python
df_vendas = pd.DataFrame(dados_vendas)
df_vendas
```

```python
df_vendas['Receita'] = df_vendas['Preço'] * df_vendas['Quantidade']
df_vendas
```

```python
#sumarizado a receeita total por produto
resumo_vendas = df_vendas.groupby('Produto')['Receita'].sum().reset_index()
resumo_vendas
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(8, 5))
plt.bar(resumo_vendas['Produto'], resumo_vendas['Receita'], color='skyblue')
plt.title('Receita Total por Produto')
```
